```python
df['Total Spend'] = df[['MntWines', 'MntFruits', 'MntMeatProducts', 'MntFishProducts', 'MntSweetProducts', 'MntGoldProds']].sum(axis=1)
```

```python
df['Total_Campain_Accepted'] = df[['AcceptedCmp1', 'AcceptedCmp2', 'AcceptedCmp3', 'AcceptedCmp4', 'AcceptedCmp5','Response']].sum(axis=1)
```

```python
df = df.rename(columns={'Total_Campain_Accepted': 'Total_Campaigns_Accepted'})
```

```python
reference_date = df['Dt_Customer'].max()
df['Customer_Tenure'] = (reference_date - df['Dt_Customer']).dt.days
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14,10))
cols = ['Income','Total Spend','Age','Customer_Tenure']
xlabels = ['Income (₹)','Total Spend (₹)','Age (years)','Customer Tenure (days)']

for ax, col, xlabel in zip(axes.flatten(), cols, xlabels):
    ax.hist(df[col], bins=30)
    ax.set_title(col)
    ax.set_xlabel(xlabel)
    ax.set_ylabel('Number of Customers')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
corr = df[['Income','Total Spend','Age','Customer_Tenure','Recency','Total_Campaigns_Accepted']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```
